# FIFA / FC 26 — retrieving data

Goal: retrieve data about FC 26's football players from Kaggle and initially check their structure
before processing in `data_pre_processing.ipynb`.

- **Source:** [rovnez/fc-26-fifa-26-player-data](https://www.kaggle.com/datasets/rovnez/fc-26-fifa-26-player-data)
- **File:** `FC26_20250921.csv`
- **Target:** `overall`

## Configuration and downloading

Things worth to mention:
- `dtype={"player_tags": str}` — enforced, so that blank values can't change column type to `float`

```python
import os
import pandas as pd
import kagglehub
from IPython.display import display

pd.set_option("display.max_rows", None)

DATASET = "rovnez/fc-26-fifa-26-player-data"

def fetch_dataset() -> pd.DataFrame:
    dataset_dir = kagglehub.dataset_download(DATASET)
    file_path = os.path.join(dataset_dir, f"FC26_20250921.csv")

    return pd.read_csv(file_path, low_memory=False, dtype={"player_tags": str})

df = fetch_dataset()
```

## Column types

We check, which columns are numerical, and which require conversion
(e.g. market value, height/weight, positional ratings saved as text like "75+2").

```python
print("\nColumns and data types:")
print(df.dtypes)
```
